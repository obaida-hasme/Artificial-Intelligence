```python
import random
file= open('/content/Arad 366 Zerind 75 Timisoara 118 Si.txt', 'r')
lines= file.readlines()
num= int(lines[0]) # number of daily transaction

def population_generator(num=num):
    population= []
    checker = 0
    while checker < num: #for loop always don't give results without error.
        tempp= ''
        for i in range(num):
            tempp+= str(random.randint(0, 1))

        if tempp != '0'* num:     #handling '00000..' chromosomes

            population.append(tempp)
            checker=checker+1 # jotogulo non full zero thik toto bar loop cholbe

        else:
          pass #000000 for this reason pass



    #print(population)
    return population

def fitness_calc(p, lines= lines, num= num):
    transact= [] #transaction and value
    for i in range(1, num+ 1):
        temp= lines[i].split()
        transact.append(dict({temp[0]: int(temp[1])})) #[{'l': 120}, {'l': 289}, {'d': 475}, {'l': 195}, {'d': 6482}, {'l': 160}, {'d': 935}]
    #print(transact)

    fitness= [] #fitness value per chromosome
    print(p) #['0011110', '0100100', '0000110', '0111100', '0011011', '1101111', '1101010']
    for i in range(num):
        lend, depo= 0, 0
        for j in range(num):
            if p[i][j]== '1':
              for k in transact[j]:
                if k== 'd':
                  depo+= transact[j][k]
                else:
                  lend+= transact[j][k]
        val= abs(depo- lend)
        fitness.append(val)
    linked= {}
    for i in range(num):
        linked[fitness[i]]= p[i]
    new= list(linked.keys())
    new.sort()
    linked_sorted= {i: linked[i] for i in new}
    fitness.sort()
    fittest, fitter= linked_sorted[fitness[0]], linked_sorted[fitness[1]]
    return fittest, fitter, linked_sorted

def crossover(chrom1, chrom2, border= random.randint(1, num)):
    new1= chrom1[0: border:]+ chrom2[border::]
    new2= chrom2[0: border:]+ chrom1[border::]
    return new1, new2

def mutation(chrom1, chrom2):
    chrom1[random.randint(0, num- 1)]= random.randint(0,1)
    chrom2[random.randint(0, num- 1)]= random.randint(0,1)
    return chrom1, chrom2


for iteration in range(10000):
    population= population_generator()
    chr1, chr2, lst= fitness_calc(population, lines= lines, num= num)
    if 0 in lst:
        print(lst[0])
        break
```
