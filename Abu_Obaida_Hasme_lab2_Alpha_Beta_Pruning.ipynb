```python
import random
std_id = input("Enter your Student ID: ")
input_string = input('Min and Max range of negative HPseparated by space:')
user_list = input_string.split()
hp_range=[]
for i in user_list:
    hp_range.append(int(i))

minimum=hp_range[0]
maximum=hp_range[1]
turns = str(std_id[:1]) #1
branches = str(std_id[2:3]) #3
Depth = 2*int(turns) #2
hp = (std_id[-2:])
main_hp = int(hp[::-1])
leaf_nodes = int(branches)**Depth
random_points = []
print('Depth and Branches ratio is',Depth,":",branches)
print(hp,main_hp)


for i in range(leaf_nodes):
    j = random.randint(minimum,maximum)
    random_points.append(j)
print("Random points:")
for i in random_points:
  print(i,end=",")

def AlphaBetaPruning(node, depth, isMaximizingPlayer, v, Alpha, Beta):
    global count
    count=0
    if depth == Depth:
        return v[node]
        print(v[node])
    if isMaximizingPlayer:
        maxEval = -10000000
        for i in range(int(branches)-1):
            value = AlphaBetaPruning(node*2 + i, depth + 1, False, v, Alpha, Beta)
            print(value)
            maxEval = max(maxEval, value)
            Alpha = max(Alpha, maxEval)
            if Beta <= Alpha:
                break
            else:
                count+=1
        return maxEval
    else :
        maxEval = 100000000
        for i in range(int(branches)-1):
            value = AlphaBetaPruning(node*2+i, depth + 1, True, v, Alpha, Beta)
            maxEval = min(maxEval, value)
            Beta = min(Beta, maxEval)
            if Alpha >= Beta:
                break
            else:
                count+=1
        return maxEval
root_value = AlphaBetaPruning(0,0,True,random_points,minimum,maximum)

print()
print("Left life(HP) of the defender after maximum damage caused by the attacker is", int(main_hp)-root_value)
print('After Alpha-Beta Pruning Leaf Node Comparisons', len(random_points)-count)
```
